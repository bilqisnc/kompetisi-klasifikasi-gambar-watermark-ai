```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Pastikan dataset sudah dimuat ke variabel 'df'
df = pd.read_csv("train.csv")  # Ganti dengan nama file CSV kamu jika berbeda

# 2. Otomatis mengambil daftar kolom yang datanya berupa angka (numerik)
numeric_features = df.select_dtypes(include=['int64', 'float64']).columns

# 3. Jalankan perulangan untuk membuat Box Plot tiap kolom
for feature in numeric_features:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x=df[feature])
    plt.title(f'Box Plot of {feature}')
    plt.show()
```

```python
import pandas as pd

# 1. Hitung Q1, Q3, dan IQR terlebih dahulu (jika belum pernah dihitung)
Q1 = df[numeric_features].quantile(0.25)
Q3 = df[numeric_features].quantile(0.75)
IQR = Q3 - Q1

# 2. Filter dataframe untuk hanya menyimpan baris yang TIDAK mengandung outliers pada kolom numerik
condition = ~((df[numeric_features] < (Q1 - 1.5 * IQR)) | (df[numeric_features] > (Q3 + 1.5 * IQR))).any(axis=1)
df_filtered_numeric = df.loc[condition, numeric_features]

# 3. Menggabungkan kembali dengan kolom kategorikal
categorical_features = df.select_dtypes(include=['object']).columns
df = pd.concat([df_filtered_numeric, df.loc[condition, categorical_features]], axis=1)
```

```python
import pandas as pd
from sklearn import datasets
from sklearn.linear_model import Lars
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# 1. Contoh Memuat Data & Melatih Model LARS
diabetes = datasets.load_diabetes()
X_train, x_test, y_train, y_test = train_test_split(
    diabetes.data, diabetes.target, test_size=0.2, random_state=42
)

lars = Lars(n_nonzero_coefs=10)
lars.fit(X_train, y_train)


# Evaluasi pada model LARS
pred_lars = lars.predict(x_test)
mae_lars = mean_absolute_error(y_test, pred_lars)
mse_lars = mean_squared_error(y_test, pred_lars)
r2_lars = r2_score(y_test, pred_lars)

# Membuat dictionary untuk menyimpan hasil evaluasi
data = {'MAE': [mae_lars], 'MSE': [mse_lars], 'R2': [r2_lars]}

# Konversi dictionary menjadi DataFrame
df_results = pd.DataFrame(data, index=['Lars'])
print(df_results)  # Gunakan print() jika dijalankan di skrip python biasa
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression, Lars
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Pastikan model LARS sudah dibuat & dilatih
lars = Lars()
lars.fit(X_train, y_train)

# Evaluasi LARS
pred_lars = lars.predict(x_test)
mae_lars = mean_absolute_error(y_test, pred_lars)
mse_lars = mean_squared_error(y_test, pred_lars)
r2_lars = r2_score(y_test, pred_lars)

# Membuat DataFrame awal
df_results = pd.DataFrame(
    {'MAE': [mae_lars], 'MSE': [mse_lars], 'R2': [r2_lars]}, index=['Lars']
)

# 2. Pastikan model Linear Regression sudah dibuat & dilatih
LR = LinearRegression()
LR.fit(X_train, y_train)

# Evaluasi Linear Regression
pred_LR = LR.predict(x_test)
mae_LR = mean_absolute_error(y_test, pred_LR)
mse_LR = mean_squared_error(y_test, pred_LR)
r2_LR = r2_score(y_test, pred_LR)

# Menambahkan hasil Linear Regression ke dalam DataFrame
df_results.loc['Linear Regression'] = [mae_LR, mse_LR, r2_LR]

# 3. Tampilkan hasil akhir (bentuk tabel biru)
df_results
```

```python
import pandas as pd
from sklearn.ensemble import (
    GradientBoostingRegressor,
)  # Jangan lupa import ini di atas
from sklearn.linear_model import LinearRegression, Lars
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Evaluasi model LARS
lars = Lars()
lars.fit(X_train, y_train)

pred_lars = lars.predict(x_test)
mae_lars = mean_absolute_error(y_test, pred_lars)
mse_lars = mean_squared_error(y_test, pred_lars)
r2_lars = r2_score(y_test, pred_lars)

df_results = pd.DataFrame(
    {'MAE': [mae_lars], 'MSE': [mse_lars], 'R2': [r2_lars]}, index=['Lars']
)


# 2. Evaluasi model Linear Regression
LR = LinearRegression()
LR.fit(X_train, y_train)

pred_LR = LR.predict(x_test)
mae_LR = mean_absolute_error(y_test, pred_LR)
mse_LR = mean_squared_error(y_test, pred_LR)
r2_LR = r2_score(y_test, pred_LR)

df_results.loc['Linear Regression'] = [mae_LR, mse_LR, r2_LR]


# ==========================================
# 3. TAMBAHKAN BAGIAN INI UNTUK GBR
# ==========================================
GBR = GradientBoostingRegressor(random_state=42)
GBR.fit(X_train, y_train)

pred_GBR = GBR.predict(x_test)
mae_GBR = mean_absolute_error(y_test, pred_GBR)
mse_GBR = mean_squared_error(y_test, pred_GBR)
r2_GBR = r2_score(y_test, pred_GBR)

df_results.loc['GradientBoostingRegressor'] = [mae_GBR, mse_GBR, r2_GBR]
# ==========================================


# 4. Tampilkan hasil akhir gabungan ketiganya
df_results
```
